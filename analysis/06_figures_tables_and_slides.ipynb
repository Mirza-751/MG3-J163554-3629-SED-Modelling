## Material for Prof. Bottacini: proton power for every η, SED data and model tables

```python
import os, json
import pandas as pd
from IPython.display import display

# (1) P_p for all eta  [1e46 erg/s, median and 68% interval]
#     P_p scales linearly with N_H/N_e, so it is computed at N_H/N_e = 1
#     and rescaled. Source, in order of preference:
#       - energ from notebook 5 (if notebook 5 was run in this session)
#       - jet_energetics_summary.json saved by notebook 5
#       - otherwise recomputed here from the chains (~1-3 min)
NH_VALUES = [1.0, 0.1, 0.01]

Pp_q = {}                                               # Pp_q[eta] = (median, -1sigma, +1sigma), N_H/N_e = 1
if 'energ' in globals():
    print("P_p from energ (notebook 5)")
    for eta in eta_grid:
        lo, med, hi = np.percentile(energ[eta]['jet_L_p_cold'], [16, 50, 84])
        Pp_q[eta] = (med, med - lo, hi - med)
elif os.path.exists('jet_energetics_summary.json'):
    print("P_p from jet_energetics_summary.json (notebook 5)")
    with open('jet_energetics_summary.json') as f:
        en_summary = json.load(f)
    for eta in eta_grid:
        Pp_q[eta] = tuple(en_summary[f"{eta:.3f}"]['P_p_NH1'])
else:
    print("P_p recomputed from the chains")
    rng = np.random.default_rng(7)
    for eta in eta_grid:
        tag = eta_tag(eta)
        d = np.load(f'chain_{tag}_7p.npz')
        samples, labels = d['samples'], list(d['labels'])
        idx = rng.choice(len(samples), size=200, replace=False)
        fm = FitModel.load_model(f'Model_Minuit_{tag}.pkl')
        jet = fm.jet_leptonic
        jet.set_external_field_transf(EXT_FIELD_FRAME)
        jet.parameters.NH_cold_to_rel_e.val = 1.0
        Pp = np.zeros(len(idx))
        for k_draw, j in enumerate(idx):
            for i, pname in enumerate(labels):
                jet.parameters.get_par_by_name(pname).val = samples[j, i]
            jet.eval()
            jet.energetic_report(verbose=False)
            Pp[k_draw] = jet.energetic_dict['jet_L_p_cold']
        lo, med, hi = np.percentile(Pp, [16, 50, 84])
        Pp_q[eta] = (med, med - lo, hi - med)
        print(f"   eta = {eta:.3f} done")

def fmt(med, ep, em):
    """median to 3 significant figures, errors with the same decimals"""
    dec = max(0, 2 - int(np.floor(np.log10(abs(med)))))
    return f"{med:.{dec}f}  (+{ep:.{dec}f} / -{em:.{dec}f})"

rows_txt, rows_num = [], []
for eta in eta_grid:
    med, em, ep = np.array(Pp_q[eta]) / 1e46
    row_txt = {'eta': f"{eta:.3f}"}
    row_num = {'eta': round(eta, 3)}
    for NH in NH_VALUES:
        row_txt[f'P_p, NH/Ne = {NH:g}'] = fmt(NH * med, NH * ep, NH * em)
        row_num[f'P_p_NH{NH:g}']      = NH * med
        row_num[f'P_p_NH{NH:g}_errp'] = NH * ep
        row_num[f'P_p_NH{NH:g}_errm'] = NH * em
    rows_txt.append(row_txt)
    rows_num.append(row_num)

Pp_table = pd.DataFrame(rows_txt).set_index('eta')
print("Cold-proton power P_p  [1e46 erg/s], one jet, median (+/- 68%)")
display(Pp_table)
pd.DataFrame(rows_num).to_csv('MG3_Pp_all_eta.csv', index=False, float_format='%.4e')

# (2) SED model at eta = 1/12: Minuit model + MCMC posterior medians
tag = eta_tag(1/12)                                     # 'eta_0p083'
fit_model = FitModel.load_model(f'Model_Minuit_{tag}.pkl')
jet = fit_model.jet_leptonic
jet.set_external_field_transf(EXT_FIELD_FRAME)

d = np.load(f'chain_{tag}_7p.npz')
samples, labels = d['samples'], list(d['labels'])
for i, pname in enumerate(labels):
    jet.parameters.get_par_by_name(pname).val = np.median(samples[:, i])
    print(f"   {pname:>11s} = {jet.parameters.get_par_by_name(pname).val:.4e}")

fit_model.eval()

nu_tot  = np.array(fit_model.SED.nu.value)              # total model (jet x EBL)
tot     = np.array(fit_model.SED.nuFnu.value)
model_table = pd.DataFrame({'nu [Hz]': nu_tot})
for comp in jet.spectral_components_list:              # single components (no EBL)
    nu_c = np.array(comp.SED.nu.value)
    f_c  = np.array(comp.SED.nuFnu.value)
    if comp.name == 'Sum' or np.all(f_c <= 0):
        continue
    name = comp.name
    f_c  = np.where(f_c > 0, f_c, 1e-300)
    model_table[name] = 10**np.interp(np.log10(nu_tot), np.log10(nu_c), np.log10(f_c))
    model_table.loc[model_table[name] < 1e-30, name] = 0.0
model_table['Total'] = tot                          # total model, as compared with the data

# SED data (as used in the fit: 15% systematics included)
t = sed_data.data
data_table = pd.DataFrame({'nu [Hz]':          np.array(t['nu_data']),
                           'nuFnu [erg/cm2/s]': np.array(t['nuFnu_data']),
                           'err [erg/cm2/s]':   np.array(t['dnuFnu_data']),
                           'upper limit':       np.array(t['UL'], dtype=bool),
                           'in fit range':      np.array(t['nu_data']) >= NU_FIT_MIN})

print("\nSED data (eta-independent)")
with pd.option_context('display.float_format', '{:.3e}'.format, 'display.max_rows', 200):
    display(data_table)
print("\nSED model at eta = 1/12, nuFnu [erg/cm2/s] (every 15th row; full table in the csv)")
with pd.option_context('display.float_format', '{:.3e}'.format, 'display.max_columns', 20):
    display(model_table[model_table['Total'] > 0].iloc[::15])

data_table.to_csv('MG3_SED_data.csv', index=False, float_format='%.5e')
model_table.to_csv('MG3_SED_model_eta0p083.csv', index=False, float_format='%.5e')
print("\nsaved: MG3_Pp_all_eta.csv, MG3_SED_data.csv, MG3_SED_model_eta0p083.csv")

# quick check: the exported model reproduces the data
fig, ax = plt.subplots(figsize=(9, 5.5))
# fixed colours, the same in every SED figure (JetSeT defaults)
colors = {'Sync': 'tab:green', 'SSC': 'tab:red', 'EC_DT': 'tab:purple', 'DT': 'tab:brown',
          'Disk': 'tab:pink', 'EC_BLR': 'tab:gray', 'EC_Disk': 'tab:olive'}
comps = [c for c in model_table.columns[1:] if c != 'Total']
for col in comps:                                       # single components
    ax.plot(model_table['nu [Hz]'], model_table[col], '--', color=colors.get(col), lw=1.2, label=col)
ax.plot(model_table['nu [Hz]'], model_table['Total'], '-', color='tab:cyan', lw=1.5, label='Total')
det = ~data_table['upper limit']
ax.errorbar(data_table['nu [Hz]'][det], data_table['nuFnu [erg/cm2/s]'][det],
            yerr=data_table['err [erg/cm2/s]'][det], fmt='o', color='tab:blue', ms=4, label='data')
ax.plot(data_table['nu [Hz]'][~det], data_table['nuFnu [erg/cm2/s]'][~det], 'v', color='tab:blue', ms=6)
ax.set_xscale('log'); ax.set_yscale('log')
ax.set_xlim(1e8, 1e27); ax.set_ylim(1e-16, 1e-10)
ax.set_xlabel(r'$\nu$ [Hz]'); ax.set_ylabel(r'$\nu F_\nu$ [erg cm$^{-2}$ s$^{-1}$]')
ax.legend(fontsize=8, ncol=2, frameon=False)
plt.show()
```

## Best-fit parameters table for the slides

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

tag = eta_tag(1/12)                                     # 'eta_0p083'
fit_model = FitModel.load_model(f'Model_Minuit_{tag}.pkl')
jet = fit_model.jet_leptonic
jet.set_external_field_transf(EXT_FIELD_FRAME)
jet.eval()

# print every parameter once, to check names and values
jet.parameters.show_pars()

# values from the model
theta    = jet.parameters.theta.val                     # deg
Gamma    = jet.parameters.BulkFactor.val
delta    = jet.get_beaming()
B        = jet.parameters.B.val                         # G
R        = jet.parameters.R.val                         # cm
R_H      = jet.parameters.R_H.val                       # cm (dissipation radius)
N        = jet.parameters.N.val                         # cm^-3
gmin     = jet.parameters.gmin.val
gmax     = jet.parameters.gmax.val
gbreak   = jet.parameters.gamma_break.val
p        = jet.parameters.p.val
p1       = jet.parameters.p_1.val

eta      = jet.parameters.accr_eff.val
M_BH     = jet.parameters.M_BH.val                      # M_sun
L_D      = jet.parameters.L_Disk.val                    # erg/s
R_in     = jet.parameters.R_inner_Sw.val                # R_S
R_out    = jet.parameters.R_ext_Sw.val                  # R_S
R_BLR_in = jet.parameters.R_BLR_in.val                  # cm
R_BLR_out= jet.parameters.R_BLR_out.val                 # cm
R_DT     = jet.parameters.R_DT.val                      # cm
T_DT     = jet.parameters.T_DT.val                      # K
tau_BLR  = jet.parameters.tau_BLR.val
tau_DT   = jet.parameters.tau_DT.val
z        = jet.parameters.z_cosm.val

# derived quantities (L_disc = L_D / (12 eta), so L_disc = L_D at eta = 1/12)
L_disc   = L_D / (12 * eta)
L_Edd    = 1.26e38 * M_BH
lam      = L_disc / L_Edd
Mdot     = L_disc / (eta * 2.998e10**2) * 3.156e7 / 1.989e33     # M_sun/yr

CHI2, DOF = 26.34, 27                                   # Minuit fit (Section 4.5.2)

print(f"delta = {delta:.2f},  L_disc = {L_disc:.3e} erg/s,  lambda = {lam:.3f},  "
      f"Mdot = {Mdot:.2f} Msun/yr,  R_diss/R_BLR,out = {R_H / R_BLR_out:.2f}")

# table rows: (symbol, description, value, unit, fixed, highlighted)
#   None = thin line between groups
def sci(x, d=1):
    """x as a.b x 10^n in mathtext"""
    n = int(np.floor(np.log10(abs(x))))
    return rf"${x / 10**n:.{d}f}\times10^{{{n}}}$"

rows_jet = [
    (r"$\theta$",             "viewing angle",        rf"${theta:.0f}$",  r"deg",              True,  False),
    (r"$\Gamma$",             "bulk Lorentz factor",  rf"${Gamma:.1f}$",  "",                  False, False),
    (r"$\delta$",             "Doppler factor",       rf"${delta:.1f}$",  "",                  False, False),
    None,
    (r"$B$",                  "magnetic field",       rf"${B:.3f}$",      "G",                 False, False),
    (r"$R$",                  "blob radius",          sci(R),             "cm",                False, False),
    (r"$R_{\rm diss}$",       "dissipation radius",   sci(R_H),           "cm",                False, True),
    None,
    (r"$N$",                  "electron density",     rf"${N:.0f}$",      r"cm$^{-3}$",        False, False),
    (r"$\gamma_{\rm min}$",   "min. Lorentz factor",  rf"${gmin:.1f}$",   "",                  True,  False),
    (r"$\gamma_{\rm break}$", "break Lorentz factor", rf"${gbreak:.0f}$", "",                  False, False),
    (r"$\gamma_{\rm max}$",   "max. Lorentz factor",  sci(gmax),          "",                  True,  False),
    (r"$p$",                  "index below break",    rf"${p:.2f}$",      "",                  True,  False),
    (r"$p_1$",                "index above break",    rf"${p1:.2f}$",     "",                  False, False),
]
rows_eng = [
    (r"$\eta$",               "accretion efficiency", rf"${eta:.3f}$",    "",                  True,  True),
    (r"$M_{\rm SMBH}$",       "black hole mass",      sci(M_BH),          r"$M_\odot$",        False, True),
    (r"$\lambda$",            "Eddington ratio",      rf"${lam:.3f}$",    "",                  False, False),
    (r"$\dot M$",             "accretion rate",       rf"${Mdot:.1f}$",   r"$M_\odot$ yr$^{-1}$", False, False),
    None,
    (r"$L_{\rm disc}$",       "disc luminosity",      sci(L_disc),        r"erg s$^{-1}$",     False, False),
    (r"$R_{\rm in}$",         "inner disc radius",    rf"${R_in:.1f}$",   r"$R_S$",            False, False),
    (r"$R_{\rm out}$",        "outer disc radius",    sci(R_out),         r"$R_S$",            False, False),
    None,
    (r"$R_{\rm BLR}$",        "BLR inner radius",      sci(R_BLR_in),      "cm",                False, False),
    (r"$R_{\rm DT}$",         "torus radius",         sci(R_DT),          "cm",                False, False),
    (r"$T_{\rm DT}$",         "torus temperature",    rf"${T_DT:.0f}$",   "K",                 False, False),
    (r"$\tau$",               "BLR, torus covering",  rf"${tau_BLR:.1f}$", "",                 True,  False),
]

# figure: two panels  |  symbol  description  value  unit
#   highlighted rows in light Unipd red, fixed parameters marked with *
UNIPD, HL, GREY = '#9b0014', '#f3dadd', '0.45'
plt.rcParams['mathtext.fontset'] = 'dejavusans'
X_SYM, X_DESC, X_VAL, X_UNIT = 0.0, 0.20, 0.80, 0.83     # column positions
GAP = 0.35                                                 # extra space between groups

panels = [('Jet', rows_jet), ('Central engine and external fields', rows_eng)]
y_end = max(1.6 + sum(r is not None for r in rows) + sum(r is None for r in rows) * GAP
            for _, rows in panels)                         # same row height in both panels

fig = plt.figure(figsize=(13, 6.2))
for k, (title, rows) in enumerate(panels):
    ax = fig.add_axes([0.01 + 0.505 * k, 0.07, 0.475, 0.92])
    ax.set_xlim(0, 1); ax.set_ylim(y_end, 0); ax.axis('off')
    ax.text(0.0, 0.5, title, fontsize=19, fontweight='bold', color=UNIPD, va='center')
    ax.plot([0, 1], [1.05, 1.05], color='0.15', lw=1.6)
    y = 1.6
    for r in rows:
        if r is None:                                      # group separator
            ax.plot([0, 1], [y - 0.5 + GAP / 2, y - 0.5 + GAP / 2], color='0.8', lw=0.8)
            y += GAP
            continue
        sym, desc, val, unit, fixed, hl = r
        if hl:
            ax.add_patch(Rectangle((-0.01, y - 0.48), 1.02, 0.96, color=HL, zorder=0))
        c = UNIPD if hl else '0.1'
        w = 'bold' if hl else 'normal'
        ax.text(X_SYM,  y, sym + (r'$^{\ast}$' if fixed else ''), fontsize=18, va='center', color=c)
        ax.text(X_DESC, y, desc, fontsize=15, va='center', color=UNIPD if hl else GREY, fontweight=w)
        ax.text(X_VAL,  y, val,  fontsize=18, va='center', ha='right', color=c)
        ax.text(X_UNIT, y, unit, fontsize=15, va='center', ha='left', color=c)
        y += 1
    ax.plot([0, 1], [y - 0.45, y - 0.45], color='0.15', lw=1.6)

fig.text(0.01, 0.025, r"$^{\ast}$ fixed in the fit", fontsize=13, color=GREY)
fig.text(0.99, 0.025, rf"Minuit best fit,  $\eta = {eta:.3f}$,  $z = {z:.3f}$,  "
         rf"$\chi^2/{{\rm dof}} = {CHI2:.1f}/{DOF}$", fontsize=13, color=GREY, ha='right')

fig.savefig('best_fit_table.png', dpi=120)               # 1560 x 744 px
plt.show()
print(f"saved best_fit_table.png")
```

## Check: L_Disk_radiative against L_D/(12 η)

```python
from jetset.model_manager import FitModel

MODEL_FILE      = 'Model_Minuit_eta_0p083.pkl'   # full path if not in this folder
EXT_FIELD_FRAME = 'blob'

fit_model = FitModel.load_model(MODEL_FILE)
jet = fit_model.jet_leptonic
jet.set_external_field_transf(EXT_FIELD_FRAME)

eta_0 = jet.parameters.accr_eff.val              # restore at the end

for eta in [0.057, 0.083, 0.15, 0.3]:
    jet.parameters.accr_eff.val = eta
    jet.eval()
    L_D    = jet.parameters.L_Disk.val
    L_rad  = jet._blob.L_Disk_radiative
    L_pred = L_D / (12 * eta)
    print(f"eta = {eta:.3f}   L_D = {L_D:.3e}   L_Disk_radiative = {L_rad:.3e}   "
          f"L_D/(12 eta) = {L_pred:.3e}   ratio = {L_rad / L_pred:.3f}")

jet.parameters.accr_eff.val = eta_0
jet.eval()
```

## Lighter copies of the figures for Overleaf

```python
import os
from PIL import Image

SRC_DIR  = 'figures'
OUT_DIR  = 'figures_overleaf'
MAX_SIDE = 2000
os.makedirs(OUT_DIR, exist_ok=True)

size_before, size_after = 0, 0
for fname in sorted(os.listdir(SRC_DIR)):
    if not fname.lower().endswith('.png'):
        continue
    src = os.path.join(SRC_DIR, fname)
    dst = os.path.join(OUT_DIR, fname)

    im = Image.open(src)
    if im.mode in ('RGBA', 'LA', 'P'):
        im = im.convert('RGBA')
        bg = Image.new('RGB', im.size, (255, 255, 255))
        bg.paste(im, mask=im.split()[-1])          # flatten transparency onto white
        im = bg
    else:
        im = im.convert('RGB')

    scale = MAX_SIDE / max(im.size)
    if scale < 1:
        im = im.resize((round(im.width * scale), round(im.height * scale)), Image.LANCZOS)

    im.save(dst, optimize=True)

    b, a = os.path.getsize(src), os.path.getsize(dst)
    size_before += b
    size_after  += a
    print(f"{fname:<45s} {b/1e6:6.2f} MB -> {a/1e6:6.2f} MB   {im.size[0]}x{im.size[1]}")

print(f"\ntotal: {size_before/1e6:.1f} MB -> {size_after/1e6:.1f} MB")
```
