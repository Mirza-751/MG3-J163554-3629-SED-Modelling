# Notebook 2: One-Zone Model — Minuit → MCMC → UltraNest

**Object:** MG3 J163554+3629 (z = 3.648)  
**Model:** One-zone SSC + EC (BLR, Disk, DT) with EBL absorption (Franceschini 2008)

This notebook runs three statistical methods in sequence:
1. **Minuit** — fast gradient minimisation to find the best-fit point
2. **MCMC** (emcee) — posterior exploration around the Minuit solution
3. **UltraNest** — nested sampling for full Bayesian posteriors + evidence

Each method builds on the previous one's result as its starting point.

```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import matplotlib.pyplot as plt
import dill
import corner
import ultranest
import ultranest.stepsampler
from scipy import stats
%matplotlib inline

from jetset.data_loader import ObsData
from jetset.minimizer import ModelMinimizer
from jetset.mcmc import McmcSampler

# Load prefit model and data from pipeline
fit_model = dill.load(open('../../results/models/prefit_model.pkl', 'rb'))
sed_data  = ObsData.load('../../results/models/sed_data.pkl')

# Use 8 threads for model evaluation
fit_model.jet_leptonic.set_num_c_threads(8)

print('Model and data loaded.')
fit_model.show_model()
```

---
## Part 1: Minuit fit

**What Minuit does:** Gradient-based minimisation of the χ² statistic. Fast and reliable for finding the best-fit point, but gives only parameter values and uncertainties (no full posterior distribution).

**Strategy:** We freeze parameters that are well-constrained by the SED shape or by prior physical arguments (θ, z, τ_DT, τ_BLR, gmin) and only fit the parameters the data can actually constrain.

```python
# ============================================================
# SET STARTING POINT (best-fit values from previous run)
# ============================================================
p = fit_model.jet_leptonic.parameters

p.gmin.val        = 2.100764
p.gmax.val        = 1.653113e4
p.p.val           = 1.590080
p.p_1.val         = 3.066185
p.gamma_break.val = 381.0763
p.N.val           = 1988.564
p.B.val           = 5.551407e-2
p.BulkFactor.val  = 16.75718
p.R_H.val         = 1.602426e18
p.M_BH.val        = 1.945859e9
p.L_Disk.val      = 1.834548e46
p.T_DT.val        = 617.6247
p.R_inner_Sw.val  = 7.269075
p.R_ext_Sw.val    = 4892.643
p.accr_eff.val    = 0.083
p.theta.val       = 3.0

# FREEZE: physically fixed or poorly constrained by data
for par in ['theta', 'theta_open', 'z_cosm', 'tau_DT', 'tau_BLR', 'gmin',
            'BulkFactor', 'N', 'gamma_break', 'p_1', 'T_DT', 'gmax', 'R_H']:
    fit_model.freeze('jet_leptonic', par)

# FREE: parameters the SED data can constrain
for par in ['B', 'M_BH', 'accr_eff', 'p', 'L_Disk', 'R_inner_Sw', 'R_ext_Sw']:
    fit_model.free('jet_leptonic', par)

# SET FIT RANGES (tight around best-fit — keeps Minuit stable)
p.B.fit_range           = [0.030,   0.090]
p.BulkFactor.fit_range  = [13.0,    21.0]
p.N.fit_range           = [1.0e3,   3.0e3]
p.gamma_break.fit_range = [2.5e2,   5.5e2]
p.p.fit_range           = [1.35,    1.80]
p.p_1.fit_range         = [2.80,    3.40]
p.R_inner_Sw.fit_range  = [3.0,     12.0]
p.R_ext_Sw.fit_range    = [1.0e3,   8.0e3]
p.T_DT.fit_range        = [4.0e2,   9.0e2]
p.L_Disk.fit_range      = [1.2e46,  3.0e46]
p.gmax.fit_range        = [8.0e3,   3.0e4]
p.M_BH.fit_range        = [8.0e8,   3.0e9]
p.accr_eff.fit_range    = [0.057,   0.300]
p.accr_eff.val_min      = 0.057
p.accr_eff.val_max      = 0.300
p.R_H.fit_range         = [1.2e18,  2.2e18]

# Verify starting point looks sensible
fit_model.eval()
pcheck = fit_model.plot_model(sed_data=sed_data)
pcheck.setlim(y_min=1E-16, x_min=1E8)
plt.title('Starting point before Minuit')
plt.show()
print('>>> Starting point verified. Running Minuit...')
```

```python
# ============================================================
# RUN MINUIT
# ============================================================
from jetset.minimizer import ModelMinimizer

model_minimizer = ModelMinimizer('minuit')
best_fit = model_minimizer.fit(
    fit_model, sed_data,
    10**10., 10**29.,
    fitname='Sum',
    repeat=1,
    use_UL=True
)

fit_model.eval()

# Save results
best_fit.save_report('../../results/models/minuit_report.pkl')
fit_model.save_model('../../results/models/minuit_model.pkl')
model_minimizer.save_model('../../results/models/minuit_minimizer.pkl')

print('\n=== Minuit best-fit parameters ===')
fit_model.show_model()

p2 = fit_model.plot_model(sed_data=sed_data)
p2.add_residual_plot(fit_model, sed_data)
p2.setlim(y_min=1E-16, x_min=1E8)
plt.suptitle('MG3 J163554+3629 — Minuit best fit')
p2.save('../../results/plots/minuit_bestfit_sed.png')
plt.show()
print(f'Beaming factor δ: {fit_model.jet_leptonic.get_beaming():.3f}')
```

---
## Part 2: MCMC with emcee

**What MCMC does:** Markov Chain Monte Carlo samples the posterior distribution P(θ|data) ∝ L(data|θ) × π(θ). Unlike Minuit, it maps out the full shape of the posterior — including degeneracies between parameters.

**emcee** uses an ensemble of walkers that move through parameter space simultaneously. We initialise them around the Minuit best-fit point.

**When to use:** When you want parameter uncertainties and correlations, but don't need Bayesian evidence (log Z).

```python
# ============================================================
# RUN MCMC (emcee via JetSeT McmcSampler)
# ============================================================
# Load the Minuit minimizer result as starting point
model_minimizer = ModelMinimizer.load_model('../../results/models/minuit_minimizer.pkl')

mcmc = McmcSampler(fit_model)

labels      = ['B', 'p', 'L_Disk', 'M_BH', 'accr_eff', 'R_inner_Sw', 'R_ext_Sw']
model_names = ['jet_leptonic'] * len(labels)
bound       = 0.5    # walkers initialised within ±50% of best-fit
ndim        = len(labels)
nwalkers    = max(2 * ndim, 20)  # at least 2*ndim walkers
nrun        = 500    # steps per walker — increase for production
burnin      = 100    # steps to discard as burn-in

sampler_mcmc, pos, prob, state = mcmc.run_sampler(
    model_minimizer,
    labels=labels,
    model_names=model_names,
    nwalkers=nwalkers,
    nrun=nrun,
    burnin=burnin,
    bound=bound,
    threads=4
)

# Save MCMC sampler
mcmc.save_sampler('../../results/models/mcmc_sampler.pkl')

print(f'MCMC complete: {nwalkers} walkers × {nrun} steps')
```

```python
# ============================================================
# MCMC DIAGNOSTICS AND CORNER PLOT
# ============================================================

# Extract flat chain (after burn-in)
flat_samples = sampler_mcmc.get_chain(discard=burnin, flat=True)
print(f'Flat chain shape: {flat_samples.shape}  ({flat_samples.shape[0]} samples)')

# Posterior medians and 1σ intervals
print('\n=== MCMC posterior summary ===')
for i, label in enumerate(labels):
    med = np.median(flat_samples[:, i])
    lo  = np.percentile(flat_samples[:, i], 16)
    hi  = np.percentile(flat_samples[:, i], 84)
    print(f'  {label:15s}: {med:.4e}  +{hi-med:.4e}  -{med-lo:.4e}')

# LaTeX labels for corner plot
latex_labels = {
    'B':          r'$B$ (G)',
    'p':          r'$p$',
    'L_Disk':     r'$L_{\rm disk}$ (erg/s)',
    'M_BH':       r'$M_{\rm BH}$ ($M_\odot$)',
    'accr_eff':   r'$\eta$',
    'R_inner_Sw': r'$R_{\rm in}$ ($R_S$)',
    'R_ext_Sw':   r'$R_{\rm ext}$ ($R_S$)',
}
clabels = [latex_labels[l] for l in labels]

fig = corner.corner(
    flat_samples,
    labels=clabels,
    quantiles=[0.16, 0.5, 0.84],
    show_titles=True,
    title_kwargs={'fontsize': 10},
    smooth=1.0
)
plt.suptitle('MG3 J163554+3629 — MCMC posterior (emcee)', y=1.02, fontsize=12)
plt.savefig('../../results/plots/mcmc_corner.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: mcmc_corner.png')
```

---
## Part 3: UltraNest — nested sampling

**What UltraNest does:** Nested sampling simultaneously estimates the posterior distribution AND the Bayesian evidence log Z. log Z is essential for **model comparison** — comparing the one-zone vs two-zone model, or different prior assumptions on T_DT.

**Key difference from MCMC:** UltraNest is slower but gives log Z (MCMC does not). Use UltraNest when you need to compare models or when the posterior has multiple modes.

**Prior choice:** We use uniform priors within physically motivated ranges derived from the Minuit fit and Sánchez Zaballa (2025) constraints.

```python
# ============================================================
# SETUP: load model, freeze all, set 6 free parameters
# ============================================================
fit_model = dill.load(open('../../results/models/minuit_model.pkl', 'rb'))
sed_data  = ObsData.load('../../results/models/sed_data.pkl')

# Freeze all parameters first
for par in fit_model.jet_leptonic.parameters.par_array:
    par.frozen = True

# Set frozen params to best-fit Minuit values
p = fit_model.jet_leptonic.parameters
p.p.val           = 1.590
p.p_1.val         = 3.066
p.gmin.val        = 2.101
p.gmax.val        = 1.653e4
p.gamma_break.val = 381.0
p.accr_eff.val    = 0.083

# 6 free parameters for UltraNest
free_params = ['B', 'BulkFactor', 'N', 'M_BH', 'L_Disk', 'T_DT']
for name in free_params:
    fit_model.jet_leptonic.parameters.get_par_by_name(name).frozen = False

# Uniform prior bounds — physically motivated ranges
prior_mins = np.array([0.020, 10.0,  500.,  5e8,  1e46, 300.])
prior_maxs = np.array([0.150, 25.0, 5000.,  5e9,  4e46, 1200.])

print(f'Free parameters: {free_params}')
print(f'Prior ranges:')
for name, lo, hi in zip(free_params, prior_mins, prior_maxs):
    print(f'  {name:15s}: [{lo:.3e}, {hi:.3e}]')
```

```python
# ============================================================
# PREPARE DATA ARRAYS FOR LIKELIHOOD
# ============================================================
# Extract data vectors once — much faster than re-reading the table each call
mask_good = sed_data.table['dnuFnu_data'] > 0  # only points with valid errors

nu_obs   = sed_data.table['nu_data'].data.astype(float)
flux_obs = sed_data.table['nuFnu_data'].data.astype(float)
flux_err = sed_data.table['dnuFnu_data'].data.astype(float)
is_ul    = sed_data.table['UL'].data.astype(bool)

print(f'Data points: {len(nu_obs)} total, {is_ul.sum()} upper limits')

def get_composite_flux(nu_target):
    """Interpolate model SED at observed frequencies."""
    fit_model.eval()
    nu_grid   = fit_model.SED.nu.value
    flux_grid = fit_model.SED.nuFnu.value
    good = np.isfinite(flux_grid) & np.isfinite(nu_grid) & (nu_grid > 0)
    nu_g, fl_g = nu_grid[good], flux_grid[good]
    idx = np.argsort(nu_g)
    return np.interp(nu_target, nu_g[idx], fl_g[idx], left=0., right=0.)

def prior_transform(cube):
    """Map unit hypercube to physical parameter space (uniform priors)."""
    return prior_mins + cube * (prior_maxs - prior_mins)

def log_likelihood(params):
    """Chi-squared likelihood with upper limit handling."""
    try:
        for name, val in zip(free_params, params):
            fit_model.jet_leptonic.set_par(name, val=val)
        model_flux = get_composite_flux(nu_obs)
        if np.any(~np.isfinite(model_flux)):
            return -1e100
        chi2 = 0.0
        for i in range(len(nu_obs)):
            if flux_err[i] <= 0:
                continue
            if is_ul[i]:
                # Upper limit: only penalise if model exceeds the UL
                if model_flux[i] > flux_obs[i]:
                    chi2 += ((model_flux[i] - flux_obs[i]) / flux_err[i])**2
            else:
                if model_flux[i] > 0:
                    chi2 += ((model_flux[i] - flux_obs[i]) / flux_err[i])**2
        return -0.5 * chi2 if np.isfinite(chi2) else -1e100
    except Exception:
        return -1e100

# Sanity check at best-fit values
best_vals = np.array([
    fit_model.jet_leptonic.parameters.get_par_by_name(n).val
    for n in free_params
])
ll_test = log_likelihood(best_vals)
print(f'logL at best-fit: {ll_test:.4f}  (should be large negative number, not -1e100)')
assert ll_test > -1e99, 'Likelihood is broken — check model evaluation.'
```

```python
# ============================================================
# RUN ULTRANEST
# ============================================================
sampler = ultranest.ReactiveNestedSampler(
    param_names = free_params,
    loglike     = log_likelihood,
    transform   = prior_transform,
    log_dir     = '../../results/ultranest_output_1zone',
    resume      = True   # resume if interrupted
)

sampler.stepsampler = ultranest.stepsampler.SliceSampler(
    nsteps             = 2 * len(free_params),
    generate_direction = ultranest.stepsampler.generate_mixture_random_direction
)

print(f'Running UltraNest: {len(free_params)} parameters, 400 live points...')
results = sampler.run(
    min_num_live_points = 400,
    max_iters           = 100000,
    show_status         = True,
    viz_callback        = False
)
sampler.print_results()
```

```python
# ============================================================
# ULTRANEST RESULTS AND PLOTS
# ============================================================
samples  = results['samples']
post_med = np.median(samples, axis=0)
post_lo  = np.percentile(samples, 16, axis=0)
post_hi  = np.percentile(samples, 84, axis=0)

print('\n=== UltraNest posterior summary ===')
for name, med, lo, hi in zip(free_params, post_med, post_lo, post_hi):
    print(f'  {name:15s}: {med:.4e}  +{hi-med:.4e}  -{med-lo:.4e}')
print(f'\nlog Z = {results["logz"]:.3f} ± {results["logzerr"]:.3f}')

# Save samples
np.save('../../results/models/ultranest_samples_1zone.npy', samples)
np.save('../../results/models/ultranest_results_1zone.npy', results, allow_pickle=True)

# Corner plot
latex_labels = {
    'B':          r'$B$ (G)',
    'BulkFactor': r'$\Gamma$',
    'N':          r'$N$ (cm$^{-3}$)',
    'M_BH':       r'$M_{\rm BH}$ ($M_\odot$)',
    'L_Disk':     r'$L_{\rm disk}$ (erg/s)',
    'T_DT':       r'$T_{\rm DT}$ (K)',
}
clabels = [latex_labels[n] for n in free_params]

fig = corner.corner(
    samples,
    labels=clabels,
    quantiles=[0.16, 0.5, 0.84],
    show_titles=True,
    title_kwargs={'fontsize': 9},
    smooth=1.0
)
# Mark Jose's M_BH value
mbh_idx = free_params.index('M_BH')
axes = np.array(fig.axes).reshape(len(free_params), len(free_params))
axes[mbh_idx, mbh_idx].axvline(1.1e9, color='red', lw=1.5, ls='--',
                                 label='Sánchez Zaballa (2025)')
axes[mbh_idx, mbh_idx].legend(fontsize=8)

plt.suptitle('MG3 J163554+3629 — UltraNest posterior (1-zone)', y=1.01, fontsize=12)
plt.savefig('../../results/plots/ultranest_corner_1zone.png', dpi=150, bbox_inches='tight')
plt.show()

# Best-fit SED using posterior median
for name, val in zip(free_params, post_med):
    fit_model.jet_leptonic.set_par(name, val=val)
fit_model.eval()

p = fit_model.plot_model(sed_data=sed_data)
p.add_residual_plot(fit_model, sed_data)
p.setlim(y_min=1E-16, x_min=1E8)
plt.suptitle('MG3 J163554+3629 — UltraNest posterior median (1-zone)')
plt.savefig('../../results/plots/ultranest_bestfit_sed_1zone.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# POSTERIOR ENVELOPE PLOT
# Shows the spread of model SEDs across the posterior
# ============================================================
n_draw = 200  # number of posterior samples to draw
idx_draw = np.random.choice(len(samples), size=n_draw, replace=False)

fig, ax = plt.subplots(figsize=(10, 6))

flux_matrix = []
nu_ref = None

for idx in idx_draw:
    for name, val in zip(free_params, samples[idx]):
        fit_model.jet_leptonic.set_par(name, val=val)
    fit_model.eval()
    nu_g   = fit_model.SED.nu.value
    flux_g = fit_model.SED.nuFnu.value
    good   = np.isfinite(flux_g) & np.isfinite(nu_g) & (nu_g > 0) & (flux_g > 0)
    if nu_ref is None:
        nu_ref = nu_g[good]
    flux_matrix.append(np.interp(nu_ref, nu_g[good], flux_g[good], left=0., right=0.))

flux_matrix = np.array(flux_matrix)
f50 = np.percentile(flux_matrix, 50, axis=0)
f16 = np.percentile(flux_matrix, 16, axis=0)
f84 = np.percentile(flux_matrix, 84, axis=0)

ax.fill_between(nu_ref, f16, f84, alpha=0.3, color='steelblue', label='posterior 68% band')
ax.loglog(nu_ref, f50, color='steelblue', lw=2, label='median model')

# Overlay data
mask_ul  = sed_data.table['UL'].astype(bool)
mask_det = ~mask_ul
ax.errorbar(sed_data.table['nu_data'][mask_det],
            sed_data.table['nuFnu_data'][mask_det],
            yerr=sed_data.table['dnuFnu_data'][mask_det],
            fmt='o', color='navy', ms=4, label='data', zorder=5)
ax.scatter(sed_data.table['nu_data'][mask_ul],
           sed_data.table['nuFnu_data'][mask_ul],
           marker='v', color='navy', s=30, zorder=5)

ax.set_xlim(1e8, 1e27)
ax.set_ylim(1e-16, 1e-10)
ax.set_xlabel(r'$\nu$ (Hz)', fontsize=13)
ax.set_ylabel(r'$\nu F_\nu$ (erg cm$^{-2}$ s$^{-1}$)', fontsize=13)
ax.set_title('MG3 J163554+3629 — UltraNest posterior envelope (1-zone)', fontsize=12)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.2)
plt.tight_layout()
plt.savefig('../../results/plots/ultranest_envelope_1zone.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved: ultranest_envelope_1zone.png')
```
