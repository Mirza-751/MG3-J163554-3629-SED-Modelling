# Two-zone model

A compact zone (SSC + EC from the BLR, disc and torus) plus an extended zone outside the torus (synchrotron + SSC),
so that the radio points can be fitted too. Builds both zones from the SED, combines them with EBL absorption and
fits the composite model with Minuit. Outputs go to `results/two_zone/`.

```python
import os
os.makedirs('results/two_zone', exist_ok=True)
os.chdir('results/two_zone')            # everything this model reads and writes lives here
```

## Data loading, SED shape, Zone 1 (compact) prefit
Object: MG3 J163554+3629 Input: SED.txt only

```python
import jetset
print('JetSeT version:', jetset.__version__)

import warnings
warnings.filterwarnings('ignore')
import numpy as np
import matplotlib.pyplot as plt

from jetset.data_loader import Data, ObsData
from jetset.sed_shaper import SEDShape
from jetset.jet_model import Jet
from jetset.obs_constrain import ObsConstrain

# 1. Load SED data
d = np.genfromtxt('../../../data/SED.txt')
d = d[d[:, 0] != 1.709E22]   # remove spurious Fermi point

data = Data(n_rows=d.shape[0])
data.set_field('x', d[:, 0])
data.set_field('y', d[:, 1])
data.set_field('dy', d[:, 2])

data.set_meta_data('z', 3.647763)
data.set_meta_data('UL_CL', 0.95)
data.set_meta_data('restframe', 'obs')
data.set_meta_data('data_scale', 'lin-lin')
data.set_meta_data('obj_name', 'MG3J163554+3629')

raw_data = ObsData(data_table=data)
sed_data = ObsData(data_table=data)

myPlot = raw_data.plot_sed()

sed_data.filter_data_set('-1', exclude=True)
sed_data.group_data(bin_width=0.15)
sed_data.add_systematics(0.15, [10.**6, 10.**29])

UL_freq = [
    1.324e25, 7.646e25, 4.188e24,
    4.188e22,
    2.417e13, 1.318e13,
]
for freq in UL_freq:
    mask = np.isclose(sed_data.table['nu_data'], freq, rtol=0.05)
    sed_data.table['UL'][mask] = True

myPlot.add_data_plot(sed_data, label='rebinned', color='red',
                      fmt='s', ms=5, mew=0.5)
myPlot.setlim(y_min=1E-16, x_min=1E8)
plt.show()

print(">>> Checkpoint 1.1: Data loaded and rebinned.")
print(f"    N points (raw)     : {d.shape[0]}")
print(f"    N points (rebinned): {len(sed_data.table)}")

# 2. SED shape fitting
my_shape = SEDShape(sed_data)
my_shape.eval_indices(silent=True)

my_shape.sync_fit(check_BBB_template=True,
                   check_host_gal_template=False,
                   use_log_par=True,
                   minimizer='lsb',
                   fit_range=[9, 16],
                   silent=True)

my_shape.IC_fit(fit_range=[16, 26], minimizer='minuit', silent=True)

p = my_shape.plot_shape_fit()
p.setlim(y_min=1E-16)
plt.show()

print(">>> Checkpoint 1.2: SEDShape characterization completed.")

# 3. Zone 1 (compact / near-zone) prefit jet
#    Location: between BLR and dust torus radius
#    Carries: Sync + SSC + EC (BLR, Disk, DT)
sed_obspar = ObsConstrain(theta=3,
                           bulk_factor=18.5,
                           B_range=[0.02, 0.08],
                           distr_e='bkn',
                           t_var_sec=7*86400,
                           nu_cut_IR=1E9,
                           SEDShape=my_shape)

zone1_jet = sed_obspar.constrain_SSC_EC_model(
    electron_distribution_log_values=False,
    EC_components_list=['EC_DT', 'EC_BLR', 'EC_Disk'],
    R_H=1.5E18,
    disk_type='MultiBB',
    silent=True
)

# Central engine values (refined)
zone1_jet.set_par('L_Disk', val=1.5817e46)
zone1_jet.set_par('M_BH', val=1.9077e9)
zone1_jet.set_par('accr_eff', val=0.083)
zone1_jet.set_par('T_DT', val=600.9)
zone1_jet.set_par('tau_DT', val=0.1)
zone1_jet.set_par('tau_BLR', val=0.1)

zone1_jet.eval()
p = zone1_jet.plot_model(sed_data=sed_data)
p.add_residual_plot(zone1_jet, sed_data)
p.setlim(y_min=1E-16, x_min=1E8)
plt.show()

print(">>> Checkpoint 1.3: Zone 1 (compact) prefit jet generated.")
print(f"    Beaming factor (delta): {zone1_jet.get_beaming()}")

# 4. Refined Zone 1 electron distribution / physics params
#    (final values validated in earlier one-zone fit)
zone1_jet.parameters.p.val = 1.3
zone1_jet.parameters.p_1.val = 2.9
zone1_jet.parameters.gamma_break.val = 400.0
zone1_jet.parameters.N.val = 1000.0
zone1_jet.parameters.B.val = 0.05

zone1_jet.parameters.gmin.val = 2.0
zone1_jet.parameters.gmax.val = 16000.0
zone1_jet.parameters.theta.val = 3.0
zone1_jet.parameters.BulkFactor.val = 15.0

zone1_jet.parameters.M_BH.val = 1.9077e9
zone1_jet.parameters.L_Disk.val = 1.5817e46
zone1_jet.parameters.accr_eff.val = 0.083
zone1_jet.parameters.T_DT.val = 600.9
zone1_jet.parameters.tau_DT.val = 0.1
zone1_jet.parameters.tau_BLR.val = 0.1
zone1_jet.parameters.R_inner_Sw.val = 10.1
zone1_jet.parameters.R_ext_Sw.val = 846.0

zone1_jet.eval()

print(">>> Checkpoint 1.4: Zone 1 refined parameters applied.")

# 5. Add R <-> R_H dependency (theta_open = 3 deg)
target_R = zone1_jet.parameters.R.val

zone1_jet.add_user_par(name='theta_open', val=3, units='deg')
zone1_jet.make_dependent_par(par='R', depends_on=['R_H', 'theta_open'],
                              par_expr='np.tan(np.radians(theta_open))*R_H')

new_RH = target_R / np.tan(3 * np.pi / 180)
zone1_jet.set_par('R_H', val=new_RH)
zone1_jet.parameters.R_H.val = 1.5e18  # final refined R_H
zone1_jet.eval()

print(">>> Checkpoint 1.5: R <-> R_H dependency added.")
print(f"    R_H = {zone1_jet.parameters.R_H.val:.4e} cm")
print(f"    R   = {zone1_jet.parameters.R.val:.4e} cm")
print(f"    R_DT = {zone1_jet.parameters.R_DT.val:.4e} cm")

# Save for Part 2
zone1_jet.save_model('zone1_jet.pkl')
sed_data.save('sed_data_rebinned.pkl')

print("\n>>> PART 1 COMPLETE.")
print("    Saved: zone1_jet.pkl, sed_data_rebinned.pkl")
print("    Proceed to Part 2: build Zone 2 and composite two-zone model.")
```

## Zone 2 (extended jet) + composite two-zone model
Zone 1 (compact): between BLR and R_DT, Sync+SSC+EC (loaded from zone1_jet.pkl) Zone 2 (extended): outside R_DT, Sync+SSC only, no EC Shared/linked params: z_cosm, theta, BulkFactor, theta_open, L_Disk, M_BH, accr_eff, T_DT, tau_DT, tau_BLR, R_inner_Sw, R_ext_Sw

```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import matplotlib.pyplot as plt

from jetset.jet_model import Jet
from jetset.data_loader import ObsData
from jetset.model_manager import FitModel
from jetset.template_2Dmodel import EBLAbsorptionTemplate

# 1. Load Zone 1 and SED data from Part 1
zone1_jet = Jet.load_model('zone1_jet.pkl')
sed_data = ObsData.load('sed_data_rebinned.pkl')

print(">>> Checkpoint 2.1: Loaded zone1_jet.pkl and sed_data_rebinned.pkl")
print(f"    Zone 1: R_H = {zone1_jet.parameters.R_H.val:.4e} cm, "
      f"R_DT = {zone1_jet.parameters.R_DT.val:.4e} cm")

# 2. Build Zone 2 (extended jet)
zone2_jet = Jet(
    name='jet_extended',
    electron_distribution='bkn',
    electron_distribution_log_values=False,
    beaming_expr='bulk_theta'
)

# Add EC machinery so Disk/BLR/DT params exist (needed for linking);
# EC emission itself will be turned off below.
zone2_jet.add_EC_component(EC_components_list=['EC_DT', 'EC_BLR', 'EC_Disk'],
                            disk_type='MultiBB')

# --- Central engine: same physical AGN ---
zone2_jet.set_par('z_cosm', val=zone1_jet.parameters.z_cosm.val)
zone2_jet.set_par('L_Disk', val=zone1_jet.parameters.L_Disk.val)
zone2_jet.set_par('M_BH', val=zone1_jet.parameters.M_BH.val)
zone2_jet.set_par('accr_eff', val=zone1_jet.parameters.accr_eff.val)
zone2_jet.set_par('R_inner_Sw', val=zone1_jet.parameters.R_inner_Sw.val)
zone2_jet.set_par('R_ext_Sw', val=zone1_jet.parameters.R_ext_Sw.val)
zone2_jet.set_par('T_DT', val=zone1_jet.parameters.T_DT.val)
zone2_jet.set_par('tau_DT', val=zone1_jet.parameters.tau_DT.val)
zone2_jet.set_par('tau_BLR', val=zone1_jet.parameters.tau_BLR.val)

# --- Geometry: viewing angle + bulk factor shared with Zone 1 ---
zone2_jet.set_par('theta', val=zone1_jet.parameters.theta.val)
zone2_jet.set_par('BulkFactor', val=zone1_jet.parameters.BulkFactor.val)

# --- theta_open and R<->R_H dependency, same opening angle ---
zone2_jet.add_user_par(name='theta_open', val=zone1_jet.parameters.theta_open.val,
                        units='deg')
zone2_jet.make_dependent_par(par='R', depends_on=['R_H', 'theta_open'],
                              par_expr='np.tan(np.radians(theta_open))*R_H')

# --- Place Zone 2 OUTSIDE R_DT ---
R_DT_zone1 = zone1_jet.parameters.R_DT.val
R_H2 = 5.0 * R_DT_zone1
zone2_jet.set_par('R_H', val=R_H2)

print(">>> Checkpoint 2.2: Zone 2 geometry/central-engine set.")
print(f"    Zone 2: R_H = {zone2_jet.parameters.R_H.val:.4e} cm "
      f"(> R_DT zone1 = {R_DT_zone1:.4e} cm)")
print(f"    Zone 2: R   = {zone2_jet.parameters.R.val:.4e} cm")

# 3. Zone 2 electron distribution and B-field
#    Lower B, much lower N (extended/diffuse population)
zone2_jet.set_par('p', val=zone1_jet.parameters.p.val)
zone2_jet.set_par('p_1', val=zone1_jet.parameters.p_1.val)
zone2_jet.set_par('gamma_break', val=zone1_jet.parameters.gamma_break.val)
zone2_jet.set_par('gmin', val=zone1_jet.parameters.gmin.val)
zone2_jet.set_par('gmax', val=zone1_jet.parameters.gmax.val)

zone2_jet.set_par('B', val=zone1_jet.parameters.B.val / 10.0)
zone2_jet.set_par('N', val=zone1_jet.parameters.N.val * 1e-4)

print(">>> Checkpoint 2.3: Zone 2 electron distribution / B-field set.")
print(f"    Zone 2: B = {zone2_jet.parameters.B.val:.4e} G "
      f"(Zone 1 B = {zone1_jet.parameters.B.val:.4e} G)")
print(f"    Zone 2: N = {zone2_jet.parameters.N.val:.4e} "
      f"(Zone 1 N = {zone1_jet.parameters.N.val:.4e})")

# 4. Turn OFF EC components for Zone 2
zone2_jet.spectral_components.EC_DT.state = 'off'
zone2_jet.spectral_components.EC_BLR.state = 'off'
zone2_jet.spectral_components.EC_Disk.state = 'off'
zone2_jet.spectral_components.DT.state = 'off'
zone2_jet.spectral_components.Disk.state = 'off'

zone2_jet.eval()

print(">>> Checkpoint 2.4: Zone 2 EC/Disk/DT components turned off.")
print("    Zone 2 active components: Sync, SSC only")

# 5. Build composite two-zone FitModel
composite_jet = FitModel(jet=zone1_jet, name='TwoZone_Model')
composite_jet.add_component(zone2_jet)

ebl_franceschini = EBLAbsorptionTemplate.from_name('Franceschini_2008')
composite_jet.add_component(ebl_franceschini)

# 6. Link shared parameters (zone_extended follows zone_leptonic)
linked_pars = ['z_cosm', 'theta', 'BulkFactor', 'theta_open',
               'L_Disk', 'M_BH', 'accr_eff',
               'T_DT', 'tau_DT', 'tau_BLR',
               'R_inner_Sw', 'R_ext_Sw']

for par in linked_pars:
    composite_jet.link_par(par_name=par, from_model='jet_extended',
                            to_model='jet_leptonic')

composite_jet.link_par(par_name='z_cosm', from_model='Franceschini_2008',
                        to_model='jet_leptonic')

print(">>> Checkpoint 2.5: Linked shared parameters (jet_extended -> jet_leptonic).")

# 7. Composite expression: sum of both zones, EBL applied to total
composite_jet.composite_expr = '(jet_leptonic + jet_extended) * Franceschini_2008'
composite_jet.eval()

print(">>> Checkpoint 2.6: Composite two-zone model built and evaluated.")
composite_jet.show_model()

# 8. Plot composite model vs data
p = composite_jet.plot_model(sed_data=sed_data)
p.add_residual_plot(composite_jet, sed_data)
p.setlim(y_min=1E-16, x_min=1E8)
plt.show()

print(">>> Checkpoint 2.7: Two-zone composite model plotted.")

# Save for Part 3
composite_jet.save_model('composite_twozone_prefit.pkl')

print("\n>>> PART 2 COMPLETE.")
print("    Saved: composite_twozone_prefit.pkl")
print("    Proceed to Part 3: Minuit fit of the composite model")
print("    (both zones), full SED including NuSTAR.")
```

## Minuit fit of the two-zone composite model
Loads composite_twozone_prefit.pkl from Part 2. Frees Zone 1's core physics params + Zone 2's N, B. This gives a sensible starting point (best-fit values) for Part 4's ultranest run.

```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import matplotlib.pyplot as plt

from jetset.model_manager import FitModel
from jetset.data_loader import ObsData
from jetset.minimizer import ModelMinimizer, Minimizer

# Workaround: JetSeT's Minimizer.corr property crashes when
# Minuit returns a degenerate covariance matrix (common with
# many free params / params near fit_range boundaries).
# Patch corr to fall back to None so fit() completes; best-fit
# *values* are unaffected, only the correlation matrix is lost.
def _safe_corr(self):
    if self.covar is None:
        return None
    cov = np.asarray(self.covar)
    if cov.ndim != 2 or cov.shape[0] != cov.shape[1]:
        print(">>> WARNING: covariance matrix malformed "
              f"(shape={cov.shape}) -- likely a parameter is "
              "pinned at a fit_range boundary. Skipping "
              "correlation matrix; best-fit values still valid.")
        return None
    v = np.sqrt(np.diag(cov))
    outer_v = np.outer(v, v)
    correlation = cov / outer_v
    correlation[cov == 0] = 0
    return correlation

Minimizer.corr = property(_safe_corr)

# 1. Load composite model and data
composite_jet = FitModel.load_model('composite_twozone_prefit.pkl')
sed_data = ObsData.load('sed_data_rebinned.pkl')

print(">>> Checkpoint 3.1: Loaded composite_twozone_prefit.pkl")

# 2. Zone 1 (jet_leptonic): freeze/free
composite_jet.jet_leptonic.set_num_c_threads(8)

composite_jet.freeze('jet_leptonic', 'theta')
composite_jet.freeze('jet_leptonic', 'theta_open')
composite_jet.freeze('jet_leptonic', 'gmax')
composite_jet.freeze('jet_leptonic', 'z_cosm')
composite_jet.freeze('jet_leptonic', 'accr_eff')
composite_jet.freeze('jet_leptonic', 'tau_DT')
composite_jet.freeze('jet_leptonic', 'tau_BLR')

composite_jet.free('jet_leptonic', 'R_H')
composite_jet.free('jet_leptonic', 'B')
composite_jet.free('jet_leptonic', 'BulkFactor')
composite_jet.free('jet_leptonic', 'N')
composite_jet.free('jet_leptonic', 'gamma_break')
composite_jet.free('jet_leptonic', 'p')
composite_jet.free('jet_leptonic', 'p_1')
composite_jet.free('jet_leptonic', 'R_inner_Sw')
composite_jet.free('jet_leptonic', 'R_ext_Sw')
composite_jet.free('jet_leptonic', 'M_BH')
composite_jet.free('jet_leptonic', 'T_DT')
composite_jet.free('jet_leptonic', 'L_Disk')
composite_jet.free('jet_leptonic', 'gmin')

composite_jet.jet_leptonic.parameters.R_H.fit_range = [5.0e+17, 3.0e+18]
composite_jet.jet_leptonic.parameters.B.fit_range = [0.010, 0.100]
composite_jet.jet_leptonic.parameters.BulkFactor.fit_range = [10.0, 25.0]
composite_jet.jet_leptonic.parameters.N.fit_range = [4.00e+02, 2.00e+03]
composite_jet.jet_leptonic.parameters.gamma_break.fit_range = [2.00e+02, 3.00e+03]
composite_jet.jet_leptonic.parameters.p.fit_range = [1.00e+00, 1.80e+00]
composite_jet.jet_leptonic.parameters.p_1.fit_range = [2.50e+00, 3.80e+00]
composite_jet.jet_leptonic.parameters.R_inner_Sw.fit_range = [3.0e+00, 1.5e+01]
composite_jet.jet_leptonic.parameters.R_ext_Sw.fit_range = [5.000e+01, 5.00e+03]
composite_jet.jet_leptonic.parameters.M_BH.fit_range = [5.0000e+08, 2.0000e+9]
composite_jet.jet_leptonic.parameters.T_DT.fit_range = [1.00000e+02, 1.000e+03]
composite_jet.jet_leptonic.parameters.L_Disk.fit_range = [1.00e+46, 5.00e+46]
composite_jet.jet_leptonic.parameters.gmin.fit_range = [1.0, 50.0]

print(">>> Checkpoint 3.2: Zone 1 (jet_leptonic) freeze/free configured.")

# 3. Zone 2 (jet_extended): free N, B only
#    (R_H, gmin, gmax, p, p_1, gamma_break stay frozen,
#     linked params controlled entirely by Zone 1)
composite_jet.jet_extended.set_num_c_threads(8)

composite_jet.freeze('jet_extended', 'gmin')
composite_jet.freeze('jet_extended', 'gmax')
composite_jet.freeze('jet_extended', 'p')
composite_jet.freeze('jet_extended', 'p_1')
composite_jet.freeze('jet_extended', 'gamma_break')

composite_jet.free('jet_extended', 'N')
composite_jet.free('jet_extended', 'B')

composite_jet.jet_extended.parameters.N.fit_range = [1.0e-03, 1.0e+01]
composite_jet.jet_extended.parameters.B.fit_range = [1.0e-04, 5.0e-02]

print(">>> Checkpoint 3.3: Zone 2 (jet_extended) freeze/free configured.")
print("    Free: N, B")
print("    Frozen: gmin, gmax, p, p_1, gamma_break, R_H")

# 4. Run Minuit
model_minimizer = ModelMinimizer('minuit')
best_fit = model_minimizer.fit(
    composite_jet, sed_data,
    10**8., 10**30.0,
    fitname='TwoZone_Sum',
    repeat=1,
    use_UL=True
)

composite_jet.eval()
composite_jet.save_model('TwoZone_final.pkl')

print("\n>>> Checkpoint 3.4: Minuit fit completed.")
composite_jet.show_model()

# 5. Plot best-fit composite model with residuals
p2 = composite_jet.plot_model(sed_data=sed_data)
p2.add_residual_plot(composite_jet, sed_data)
p2.setlim(y_min=1E-16, x_min=1E8)
plt.show()

print(f"\nBeaming factor (Zone 1): {composite_jet.jet_leptonic.get_beaming()}")
print(f"Beaming factor (Zone 2): {composite_jet.jet_extended.get_beaming()}")

print("\n>>> PART 3 COMPLETE.")
print("    Saved: TwoZone_final.pkl")
print("    Proceed to Part 4: ultranest fit (both zones, radio points")
print("    included in fit_mask).")
```
